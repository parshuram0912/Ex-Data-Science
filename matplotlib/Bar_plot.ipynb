```python
import matplotlib.pyplot as plt

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

# Use parentheses to pass arguments into the functions
plt.bar(years, runs)
plt.xlabel("Years")
plt.ylabel("Runs that score") 

# Use parentheses to execute the show function
plt.show()
```

```python
import numpy as np
x=np.arange(len(years))
print(x)
width=0.25
```

```python
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

plt.bar(x-width,kohli,width=width,label="kolhi")
plt.bar(x,sachin,width=width,label="sachin")
plt.bar(x+width,sehwag,width=width,label="sehwag")

plt.xlabel=("years")
plt.ylabel=("runs")
plt.label=("shows the runs in the cricket")

plt.xticks(x,years)  #here are the main part are xticks replace the x value are in the years
plt.legend()
plt.show()
```
