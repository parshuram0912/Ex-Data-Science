```python
import matplotlib
import matplotlib.pyplot as plt
plt.plot([1, 2, 3], [4, 5, 6])
plt.show()
```

```python
cars=["Alto","celerio","swift","Astar","brezza","victriou"]
sales=[2000,6000,4500,6400,2300,4800]
plt.plot(cars,sales)
plt.xlabel("cars module")
plt.ylabel("unit")
plt.title("sales comparision")
plt.show()
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli,'ro--', label="Virat Kohli") #this is the line with string r for red color and the -- for line style
plt.plot(years, sehwag, label="Virender Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend()
plt.show()
```

```python
plt.plot(cars, sales, label="Car Sales", marker='o', color='purple')
plt.xlabel("Car Models")
plt.ylabel("Units Sold")
plt.title("Sales by Car Model")
plt.legend()
plt.show()
```
